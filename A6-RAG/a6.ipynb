### Contextual

```python
import asyncio
from langchain_groq import ChatGroq
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
async def enrich_chunk(chunk_text: str, full_doc_text: str, title: str) -> str:
    """Generates a situational prefix for a chunk."""
    prompt = f"""
    <document>
    {full_doc_text[:4000]} 
    </document>
    
    Here is a specific chunk from the document:
    <chunk>
    {chunk_text}
    </chunk>
    
    Provide a 1-sentence context situating this chunk within the overall document.
    Format: "This chunk from [title] discusses [explanation]."
    """
    
    response = await llm.ainvoke(prompt)
    context_prefix = response.content.strip()
    return f"{context_prefix}\n\n{chunk_text}"
```

```python
async def process_all_chunks(chunks, full_text, title):
    enriched_docs = []
    print(f"Enriching {len(chunks)} chunks... please wait.")
    
    for chunk in chunks:
        # Create the contextualized text
        contextualized_text = await enrich_chunk(chunk.page_content, full_text, title)
        
        # Preserve original metadata but update the content
        new_doc = Document(
            page_content=contextualized_text,
            metadata=chunk.metadata
        )
        enriched_docs.append(new_doc)
    
    return enriched_docs
```

```python
full_text = "\n\n".join([doc.page_content for doc in raw_docs])
```

```python
enriched_chunks = await process_all_chunks(
    chunks=documents, 
    full_text=full_text, 
    title="Chapter 11: Information Retrieval"
)
```

```python
embeddings_gpu = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2", 
    model_kwargs={'device': 'cuda'}
)
```

```python
contextual_db = Chroma.from_documents(
    documents=enriched_chunks,
    embedding=embeddings_gpu,
    persist_directory="./chroma_contextual"
)
```

```python
contextual_retriever = contextual_db.as_retriever(search_kwargs={"k": 3})

contextual_qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=contextual_retriever,
    return_source_documents=True
)
```

```python
query = "What is an inverted index?"

#naive retrieval
naive_results = vector_db.similarity_search(query, k=1)
print(f"NAIVE MATCH: {naive_results[0].page_content[:200]}...")

#contextual retrieval
contextual_results = contextual_db.similarity_search(query, k=1)
print(f"CONTEXTUAL MATCH: {contextual_results[0].page_content[:200]}...")
```

```python
import json


contextual_retriever = contextual_db.as_retriever(search_kwargs={"k": 3})

contextual_qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=contextual_retriever,
    return_source_documents=True
)

print("Starting evaluation loop for 20 questions...")

for i, item in enumerate(pair_20):
    question = item["question"]
    print(f"[{i+1}/20] Processing: {question}")
    
    try:
        naive_response = qa_chain.invoke({"query": question})
        item["naive_rag_answer"] = naive_response["result"]
        
        contextual_response = contextual_qa_chain.invoke({"query": question})
        item["contextual_retrieval_answer"] = contextual_response["result"]
    except Exception as e:
        print(f"Error processing question {i+1}: {e}")
        item["naive_rag_answer"] = "ERROR"
        item["contextual_retrieval_answer"] = "ERROR"

with open('rag_results.json', 'w') as f:
    json.dump(pair_20, f, indent=4)

print("Done! All answers generated and saved to 'rag_results.json'.")

```

```python
import json
import pandas as pd
from rouge_score import rouge_scorer


with open('rag_results.json', 'r') as f:
    data = json.load(f)

scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

results_table = []

for entry in data:
    question = entry['question']
    gt = entry['ground_truth_answer']
    
    for model_name, column in [("Naive RAG", "naive_rag_answer"), ("Contextual Retrieval", "contextual_retrieval_answer")]:
        candidate = entry[column]
        
        # Handle "I don't know" cases 
        scores = scorer.score(gt, candidate)
        
        results_table.append({
            "Question": (question[:45] + '...') if len(question) > 45 else question,
            "Model": model_name,
            "ROUGE-1 (F1)": round(scores['rouge1'].fmeasure, 4),
            "ROUGE-2 (F1)": round(scores['rouge2'].fmeasure, 4),
            "ROUGE-L (F1)": round(scores['rougeL'].fmeasure, 4)
        })

df = pd.DataFrame(results_table)
print(df.to_string(index=False))
```

## Understanding the results

<b>Low Overall Scores:</b> Most ROUGE-1 scores hover between 0.10 and 0.35, while ROUGE-2 (bigram overlap) is frequently below 0.10. This suggests that while the models capture the general concepts, they do not perfectly mimic the concise phrasing of the ground truth.

<b>Naive RAG’s Edge in Precision:</b> In several instances, such as the "Primary Goal of IR" and "Role of the Generator," Naive RAG achieves higher ROUGE-1 and ROUGE-L scores than Contextual Retrieval. This often happens when the Naive model provides a shorter answer that aligns more closely with the reference length.

Some findings interpreted from the experiment are :

1. ROUGE-2 (F1) "Zero" Scores: There are multiple instances (e.g., "Inverted Index" and "Knowledge Grounding" for Naive RAG) where the ROUGE-2 score is 0.0000. This indicates that while individual words might match (ROUGE-1), the model failed to capture any exact two-word sequences from the ground truth.

2. Contextual Retrieval Successes: Contextual Retrieval significantly outperforms Naive RAG on the "IDF" question, nearly doubling the ROUGE-1 score (0.2969 vs 0.1758) and ROUGE-2 score (0.1587 vs 0.0444). This suggests that for certain technical definitions, the contextual approach retrieved more precise terminology.

3. Model Failures (MRR and Knowledge Grounding): For "Mean Reciprocal Rank (MRR)" and "Knowledge Grounding," the Contextual Retrieval model scored 0.0000 across all metrics. Cross-referencing this with the raw data shows that the model responded with "I don't know," leading to zero overlap with the factual ground truth.
